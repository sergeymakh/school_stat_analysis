#### Образовательные учреждения города Москвы
Образовательные учреждения города Москвы: https://data.mos.ru/opendata/2263?isDynamic=false  
Файл: data-2263-2025-08-25.csv

```python
df = pd.read_csv('data-2263-2025-08-25.csv', sep=';', skiprows=1)
df = df.drop('Unnamed: 26', axis=1)
```

```python
# df.head()
```

```python
new_column_names = {
    'global_id': 'global_id',
    'Тип образовательной организации': 'institution_type',
    'Статус организации': 'organization_status',
    'КПП': 'kpp',
    'Лицензирование и аккредитация': 'licensing_and_accreditation',
    'ОГРН': 'ogrn',
    'Организационно-правовая форма': 'legal_form',
    'Подчинение': 'subordination',
    'ФИО руководителя': 'director_fullname',
    'Юридический адрес': 'legal_address',
    'Контактный телефон': 'contact_phone',  # Русско-английское переименование
    'Адрес электронной почты': 'email',      # Русско-английское переименование
    'Сайт': 'website',
    'Образовательные программы': 'education_programs',
    'Образовательные услуги': 'education_services',
    'Количество обучающихся по ОО': 'students_count',
    'Контингент обучающихся дошкольников по ОО': 'preschool_students_count',
    'Код ЕКИС': 'ekis_code',
    'Полное наименование': 'full_name',
    'Сокращённое наименование': 'short_name',
    'Номер': 'number',
    'Адреса учреждения': 'addresses',
    'ИНН': 'inn',
    'Кружки и секции': 'clubs_and_sections',
    'geoData': 'geodata',
    'geodata_center': 'geodata_center'
}

df.rename(columns=new_column_names, inplace=True)
```

```python
with engine.connect() as conn:
    conn.execute(text("TRUNCATE TABLE schooldata.schools"))
    conn.commit()
conn.close()
df.to_sql(name='schools', con=engine, schema='schooldata', if_exists='append', index=False)
```

#### Победители олимпиад
Победители олимпиад: https://data.mos.ru/opendata/1093?isDynamic=false  
Файл: data-1093-2025-09-01.csv

```python
df = pd.read_csv('data-1093-2025-09-01.csv', sep=';', skiprows=1)
df = df.drop('Unnamed: 10', axis=1)
```

```python
new_column_names = {
    'Полное официальное наименование': 'full_name',
    'Сокращенное наименование': 'short_name',
    'Тип олимпиады': 'competition_type',
    'Этап': 'stage',
    'Класс': 'grade',
    'Наименование олимпиады': 'competition_name',
    'Профиль олимпиады': 'subject_profile',
    'Статус (победитель / призёр)': 'status',
    'Год участия в олимпиаде': 'participation_year'
}

df.rename(columns=new_column_names, inplace=True)
```

```python
with engine.connect() as conn:
    conn.execute(text("TRUNCATE TABLE schooldata.olimpic_winners"))
    conn.commit()
conn.close()
df.to_sql(name='olimpic_winners', con=engine, schema='schooldata', if_exists='append', index=False)
```
